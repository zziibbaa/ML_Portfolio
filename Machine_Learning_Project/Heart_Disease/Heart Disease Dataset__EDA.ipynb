```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('processed.cleveland.data', header=None)

data.columns = [
    'age','sex','cp','trestbps','chol','fbs','restecg',
    'thalach','exang','oldpeak','slope','ca','thal','DISEASE'
]

data.replace('?', np.nan, inplace=True)
data.dropna(inplace=True)

data['DISEASE'] = (data['DISEASE'] > 0).astype(int)
```

```python
sns.countplot(x='DISEASE', data=data)
```

```python
num_cols = ['age','trestbps','chol','thalach','oldpeak']

for col in num_cols:
    sns.histplot(data[col], kde=True)
    plt.title(col)
    plt.show()
```

```python
for col in num_cols:
    sns.boxplot(x='DISEASE', y=col, data=data)
    plt.title(f"{col} vs Disease")
    plt.show()
```

```python
cat_cols = ['sex','cp','fbs','restecg','exang','slope','ca','thal']
for col in cat_cols:
    sns.countplot(x=col, hue='DISEASE', data=data)
    plt.title(f"{col} vs Disease")
    plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap='coolwarm')
```

```python
for col in ['cp','ca','thal']:
    print(pd.crosstab(data[col], data['DISEASE'], normalize='index'))
```
